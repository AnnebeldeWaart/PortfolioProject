### Hoofdscenario:

- **Doel:** API wordt succesvol ingeladen en de data wordt in het `csv` bestand gezet.
- **Input:** Correcte API-URL en geldige API-key en ID.
- **Verwacht resultaat:** Data wordt ingeladen zonder fouten.

### Scenario 1
1. **Situatie:** Om er voor te zorgen dat de data van de API juist ontvangen wordt, moet ik dit goed testen en opvangen. Omdat het om 1000 regels gaat wil ik de error goed kunnen opvangen voor dat ik er verder mee aan de slag ga.

2. **Expected:** In het geval dat de code goed werkt zou het aantal opgeslagen vacatures moeten printen.


### Scenario 2

1. **Situatie:** Nu worden de aantal opgeslagen vacatures geprint. Maar het belangerijke is hoe we de error gaan opvangen als er wat mislukt met het inladen van de `API`. Dit is namelijk een veel voorkomende error en die wil ik zo goed mogelijk opvangen.

2. **Expected:** De error wordt opgevangen met de foutmelding en error code 404 en bij welke pagina het mis gaat.




### Scenario 3

1. **Situatie:** Er is nu de fout melding 404. Het is handig om uit te leggen hoe je hier mee om kan gaan.

**Observatie:**  Uit deze vergelijking blijkt duidelijk welke categorieën zijn aangepast, vervangen of toegevoegd. Categorieën die eerst `unknown` waren, zijn nu vervangen door waarden van de functietitel.

**Observatie:** Zoals te zien, zijn  `unknown` & `Vacatures Ander of Algemeen` niet maar aanwezig, en zijn deze netjes opgevult.

## Test 3: Veel NaN waardes tussen de salaris data.

**User Story Salarisinformatie verwerken:**     
Als data-analist wil ik dat salarissen correct worden opgeschoond, zodat mijn berekeningen betrouwbaar zijn.

### Hoofdscenario:

- **Doel:** Alle `salary` waardes opvullen, maar er voor zorgen dat de waardes met het opvullen realtisch blijven.
- **Input:** De `salary` kolom.
- **Expected:** Dat de `salary` kolom zo veel mogelijke is gevuld wordt met data.

### Scenario 1

1. **Situatie:** Ik wil testen hoe de kwaliteit van `salary_min` en `salary_max` zijn.

2. **Expected:** Veel ontbrekende waarden in de salaris kolom. Zowel bij `salary_min` als `salary_max`.




**Observatie:**
Zoals te zien in het screenshot, zijn er veel regels waar `salary_min`& `salary_max`NaN waarde hebben.
Ik moet een manier vinden en testen om te kijken of ik de salarisen kan opvullen.

### Scenario 2

1. **Situatie:** Er zijn veel verschillende categorieën, daardoor wordt de mediaan per categorie gebruikt om de ontbrekende waarden in te vullen in plaats van de mediaan van de gehele dataset.

2. **Expected:** Voor elke vacature wordt het gemiddelde salaris `salary_row_mean` berekend op basis van `salary_min` en `salary_max`. Het gemiddelde salaris per categorie `salary_mean` bepaald en waar nodig aangevuld met het gemiddelde.



**Observatie:**  
Na het vullen van de NaN-waarden met het gemiddelde per categorie, zijn de meeste ontbrekende waarden ingevuld.  
Sommige categorieën bevatten te weinig data, waardoor er nog steeds NaN-waarden aanwezig zijn.  
Het zou niet verstandig zijn om deze resterende NaN-waarden op te vullen met de mediaan van de hele dataset, omdat de salarissen per vacature sterk kunnen verschillen. Dit zou de nauwkeurigheid van de data verminderen.

### Scenario 3

1. **Situatie:** Er zijn dus nogsteeds categorieën die geen salaris bevatten na het vullen van de NaN-waarden.

2. **Expected:** De kolommen `salary_min` en `salary_max` worden eerst omgezet naar numeriek, zodat eventuele fouten of niet-numerieke waarden als NaN worden behandeld.


Toen kwam het volgende er uit, maar deze uitkomst bleek nog niet correct te zijn.



Ik berekende de `Salary_min` en `Salary_max` in het vorige code blok, maar in het code blok daarna deed ik perongeluk nog een keer `df["salary_max"] = pd.to_numeric(df["salary_max"], errors="coerce")`. Daardoor maak ik weer NaN van de net opgevulde waardes.

Nu komen de juiste waardes naar voren:



**Observatie:** Hier kan je zien welke `category` geen salary info hebben. Ook kunnen we zien hoeveel regels de category bevat.     
Onderin kan je duidelijk zien hoeveel van de vacatures wel en geen salaris hebben. Helaas kan ik met deze data niks meer kwa salaris informatie, omdat de data te incompleet is. Maar gelukkig heb ik wel voor veel NaN waardes een oplossing gevonden. Met het verbeter voorstel ga ik hier nog verder op in.
